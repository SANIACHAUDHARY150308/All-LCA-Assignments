```python
#Create two matrices using nested lists
matrix1 = [
    [1,3,5],
    [7,9,1]
]
matrix2 = [
    [2,4,6],
    [8,2,4]
]
#Initialize a result matrix with the same size
result=[
    [0,0,0],
    [0,0,0]
]
#perform matrix addition
for i in range(len(matrix1)):                               #loop through rows
   for j in range(len(matrix2[0])):                           #loop through columns
     result[i][j] = matrix1[i][j] + matrix2[i][j]
#Display the result
print("Matrix 1:")
for row in matrix1:
    print(row)

print("\nMatrix 2:")
for row in matrix2:
    print(row)

print("\nSum of Matrix 1 and Matrix 2:")
for row in result:
    print(row)
```
