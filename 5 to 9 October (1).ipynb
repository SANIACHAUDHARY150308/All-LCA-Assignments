```python
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

physics = float(input("Enter Physics marks: "))
maths = float(input("Enter Maths marks: "))
chemistry = float(input("Enter Chemistry marks: "))

total = physics + maths + chemistry
percentage = total / 3

print("Student Name:", name)
print("Roll Number:", roll_no)
print("Percentage:", percentage)

if physics >= maths and physics >= chemistry:
    print("Highest: Physics")
elif maths >= physics and maths >= chemistry:
    print("Highest: Maths")
else:
    print("Highest: Chemistry")

if physics <= maths and physics <= chemistry:
    print("Lowest: Physics")
elif maths <= physics and maths <= chemistry:
    print("Lowest: Maths")
else:
    print("Lowest: Chemistry")
```

```python
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
num = int(input("Enter a number: "))

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reverse =", rev)
```

```python
num = int(input("Enter a number: "))

original = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if original == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
numbers = []

for i in range(20):
    num = int(input("Enter a number: "))
    numbers.append(num)

print("Similar elements:")

for i in range(20):
    if numbers.count(numbers[i]) > 1:
        print(numbers[i], "at index", i)

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even =", even)
print("Odd =", odd)

positive = 0
negative = 0

for num in numbers:
    if num > 0:
        positive = positive + 1
    elif num < 0:
        negative = negative + 1

print("Positive =", positive)
print("Negative =", negative)
```
